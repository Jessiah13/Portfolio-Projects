```python
# import needed libraries: 

import os
import glob
import shutil
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.edge.service import Service
from selenium.webdriver.edge.options import Options
import time

from selenium.webdriver.support.ui import Select
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


from time import sleep
from PIL import Image
```

```python
# Assign file path: 

file_path = "C:\\Users\\Jessiah\\Documents\\Python\\Python Portfolio Projects\\PD Translator\\photos\\"
image_folder_path = "C:\\Users\\Jessiah\\Downloads\\"

# place files with the specify format into a list
files = glob.glob(file_path + "*.jpg")
```

```python
# open browswer and load webpage
driver = webdriver.Chrome()
url = "https://translate.yandex.com/"

driver.get(url)

# wait for webpage to laod
sleep(2)
```

```python
# select the image option

image_button = driver.find_element(By.XPATH, '//*[@data-text="Images"]')
ActionChains(driver).move_to_element(image_button).click(image_button).perform()

# wait for page to load
sleep(2)
```

```python
# Get the file input to upload files
file_input = driver.find_element(By.CSS_SELECTOR, 'input[type="file"]')
```

```python
# Loop through the list of files. Upload, translate and download them

for file in files: 
    file_input.send_keys(file)
    # wait for upload: 
    sleep(3)

    # Select english as the disired language
    to_lang = driver.find_element(By.ID, "dstLangButton")
    ActionChains(driver).move_to_element(to_lang).click(to_lang).perform()
    # wait 
    sleep(1)
    eng = driver.find_element(By.XPATH, '//*[@aria-label="English"]')
    ActionChains(driver).move_to_element(eng).click(eng).perform()
    #wait again: 
    sleep(2)

    # find and select download button: 
    download = driver.find_element(By.ID, "downloadButton")
    ActionChains(driver).move_to_element(download).click(download).perform()
    # wait again
    sleep(2)

    #clear file and start again 
    clear = driver.find_element(By.XPATH, '//*[@data-action="resetImage"]')
    ActionChains(driver).move_to_element(clear).click(clear).perform()
    # wait
    sleep(2)
    
```

```python
# close the borwser
driver.quit()
```

```python
# create an empty list
images = []

#rename the files to what they will be after translation. 
for file in files: 
    file_name = file.split('\\')
    file_name = file_name[-1].split('.')
    file_name = file_name[0] + ".translated.jpg"
    print(file_name)
    # add to list
    images.append(file_name)
```

```python
# create empty list
image_list = []

# loop thought each file, convert to RGB and append to list.
for image in images: 
    image_path = os.path.join(image_folder_path, image)
    img = Image.open(image_path)
    img = img.convert('RGB')
    image_list.append(img)
```

```python
# merge all imgages into a single PDF file.
output_pdf_path = 'C:\\Users\\Jessiah\\Documents\\Python\\Python Portfolio Projects\\PD Translator\\output files\\merged_images_output.pdf'
if image_list:
    image_list[0].save(output_pdf_path, save_all=True, append_images=image_list[1:])
```
